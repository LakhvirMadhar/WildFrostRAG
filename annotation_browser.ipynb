# WildFrostRAG Experiment Annotation Browser

This notebook provides an interactive interface for browsing and annotating experiments.

## Features
- **Auto-discovery**: Automatically finds all experiments from the registry
- **Filtering**: Filter by run number, experiment type, or retriever
- **Preview**: See experiment details before opening
- **Modular**: Works with any experiment type - no code changes needed for new retrievers

## Usage
1. Run the cell below to launch the browser
2. Use the filters to find your experiment
3. Click an experiment to see its details
4. Click "Open Annotation GUI" to start annotating

```python
from src.gui import browse_experiments

# Launch the experiment browser
browser = browse_experiments()
```

---

## Get Annotation Summary

After annotating, you can get a summary of your progress:

```python
# Get the current annotation GUI (after selecting and opening an experiment)
gui = browser.get_annotation_gui()

if gui:
    summary = gui.get_validation_summary()
    print("Validation Summary:")
    print(f"  Pass: {summary['pass']}")
    print(f"  Fail: {summary['fail']}")
    print(f"  Unvalidated: {summary['unvalidated']}")
    print(f"  Total: {sum(summary.values())}")
else:
    print("No annotation GUI active. Select an experiment first.")
```

## List All Experiments Programmatically

```python
from src.gui import ExperimentRegistry

registry = ExperimentRegistry()

print(f"Current run: {registry.get_current_run()}")
print(f"All runs: {registry.get_all_runs()}")
print(f"Retrievers: {registry.get_retrievers()}")
print()

# List all retrieval experiments
print("Retrieval Experiments:")
for exp in registry.get_experiments(experiment_type='retrieval'):
    print(f"  - {exp['id']}: {exp.get('retriever_type')} ({exp.get('total_queries')} queries)")

print()
print("Generation Experiments:")
for exp in registry.get_experiments(experiment_type='generation'):
    print(f"  - {exp['id']}: {exp.get('system_prompt_version')} (source: {exp.get('retrieval_reference')})")
```
